# 1. Dynamic Programming (DP) ⭐⭐⭐ **HIGHEST PRIORITY**

## 🎯 Why Airbnb Loves DP
- Real-world optimization (pricing, scheduling, resource allocation)
- Tests problem decomposition skills
- Common in system design + coding rounds

## 💡 Core DP Pattern Recognition

**Key Questions to Ask:**
1. Can I break this into smaller subproblems?
2. Do subproblems overlap? (If yes → DP)
3. What's the optimal substructure? (Optimal solution contains optimal sub-solutions)

**DP Template:**
```python
# 1. Define state: dp[i] or dp[i][j] represents what?
# 2. Find recurrence: How to build current from previous states?
# 3. Base cases: Smallest subproblems
# 4. Build bottom-up or memoize top-down
```


### Pattern 1: Linear DP (1D)

**Formula:** `dp[i] = f(dp[i-1], dp[i-2], ...)`

**When to Use:**
- Sequential decisions (rob house, climb stairs)
- Optimization along a sequence

**Key Insight:** Often can optimize space to O(1) using variables instead of array

**Problems:**
| Problem | File | Key Pattern |
|---------|------|-------------|
| House Robber (LC #198) | `house_robber.py` | `dp[i] = max(dp[i-1], dp[i-2] + nums[i])` |
| House Robber II (LC #213) | `house_robber_ii.py` | Circular array variant |
| House Robber III (LC #337) | `house_robber_iii.py` | Tree DP (bottom-up) |
| Maximum Subarray (LC #53) | `maximum_subarray.py` | Kadane's: `max(nums[i], curr + nums[i])` |
| Coin Change (LC #322) | `coin_change.py` | `dp[amount] = min(dp[amount], 1 + dp[amount-coin])` |
| Maximum Profit Job Scheduling (LC #1235) | `maximum_profit_job_scheduling.py` | DP + Binary Search |

**🎓 Interview Tips:**
- Start with brute force recursion, then add memoization
- Identify: "What's the optimal solution up to position i?"
- Look for space optimization (O(1) for linear DP)
- Mention time/space trade-offs


### Pattern 2: 2D DP (Matrix)

**Formula:** `dp[i][j] = f(dp[i-1][j], dp[i][j-1], dp[i-1][j-1])`

**When to Use:**
- Two sequences/strings comparison
- Matrix traversal with constraints
- Edit distance, longest common subsequence

**Key Insight:** Often can optimize space to O(min(m,n)) using rolling array

**Problems:**
| Problem | File | Key Pattern |
|---------|------|-------------|
| Edit Distance (LC #72) | `edit_distance.py` | `dp[i][j] = min(insert, delete, replace)` |
| Maximal Square (LC #221) | `maxsquare.py` | `dp[i][j] = min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1]) + 1` |
| K Edit Distance | `k_edit_distance.py` | Edit distance with constraint |

**🎓 Interview Tips:**
- Define state clearly: "What does dp[i][j] represent?"
- Draw the DP table for first example
- Mention space optimization possibility
- Handle edge cases (empty strings, single characters)


### Pattern 3: DP with State Machine

**Formula:** `dp[i][state] = optimal value at position i with state`

**When to Use:**
- Multiple states (holding stock, not holding)
- State transitions with constraints
- House Robber with circular constraint

**Problems:**
| Problem | File | States |
|---------|------|--------|
| House Robber II | `house_robber_ii.py` | Rob first vs skip first |
| House Robber III | `house_robber_iii.py` | Rob current vs skip current (tree) |

**🎓 Interview Tips:**
- Identify all possible states
- Define transitions between states
- Often can optimize to O(1) space per state


### Pattern 4: DP with Memoization (Top-Down)

**When to Use:**
- Natural recursive structure
- Not all states are needed
- Easier to think recursively

**Problems:**
| Problem | File | Why Memoization |
|---------|------|-----------------|
| Shopping Offers (LC #638) | `shopping_offers.py` | State = needs vector, sparse states |
| Pyramid Transition Matrix (LC #756) | `pyramid_transition_matrix.py` | Backtracking + memoization |
| House of Cards (LC #2189) | `house_of_cards.py` | Counting ways to partition |

**🎓 Interview Tips:**
- Use `@lru_cache` or dictionary for memoization
- Convert recursive parameters to hashable types (tuples)
- Mention that bottom-up might be more space-efficient

# 2. Graph Algorithms ⭐⭐⭐ **CRITICAL**

## 🎯 Why Airbnb Loves Graphs
- Dependencies in booking systems
- Route planning and recommendations
- Social network features (travel buddies)
- Topological sorting for task scheduling

## 💡 Graph Pattern Recognition

**Key Questions:**
1. Are there dependencies/ordering? → Topological Sort
2. Need shortest path? → BFS (unweighted) or Dijkstra (weighted)
3. Need to explore all nodes? → DFS or BFS
4. Need to find cycles? → DFS with state tracking


### Pattern 1: Topological Sort

**When to Use:**
- Dependencies/ordering problems
- Course scheduling
- Build systems with dependencies
- Alien dictionary (character ordering)

**Algorithm (Kahn's - BFS-based):**
```python
# 1. Build graph and in-degree array
# 2. Add all nodes with in-degree 0 to queue
# 3. While queue not empty:
#    - Pop node, add to result
#    - Decrease in-degree of neighbors
#    - If neighbor in-degree becomes 0, add to queue
# 4. If result length == total nodes: valid ordering
#    Else: cycle exists
```

**Problems:**
| Problem | File | Key Insight |
|---------|------|-------------|
| Alien Dictionary (LC #269) | `aliendictionary.py` | Build graph from word pairs, topological sort |
| Minimum Vertices to Reach All Nodes | `minimum_vertices_reach_all.py` | Find nodes with in-degree 0 |

**🎓 Interview Tips:**
- Recognize dependency pattern immediately
- Build graph: adjacency list + in-degrees
- Use BFS/queue for topological sort
- Always handle cycles (return empty/False if cycle detected)
- Time: O(V+E), Space: O(V+E)


### Pattern 2: DFS/BFS on Graphs

**When to Use:**
- Finding paths
- Connected components
- Cycle detection
- Exploring all possibilities

**DFS Template:**
```python
def dfs(node, visited):
    visited.add(node)
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor, visited)
```

**BFS Template (Shortest Path):**
```python
from collections import deque
queue = deque([(start, 0)])  # (node, distance)
visited = {start}
while queue:
    node, dist = queue.popleft()
    if node == target: return dist
    for neighbor in graph[node]:
        if neighbor not in visited:
            visited.add(neighbor)
            queue.append((neighbor, dist + 1))
```

**Problems:**
| Problem | File | Algorithm |
|---------|------|-----------|
| Word Search (LC #79) | `word_search.py` | DFS with backtracking |
| Word Search II (LC #212) | `word_search_ii.py` | DFS + Trie |
| Making a Large Island (LC #827) | `making_a_large_island.py` | DFS to label islands, then try connecting |
| Shortest Path in Binary Matrix (LC #1091) | `shortest_path_binary_matrix.py` | BFS (8-directional) |
| Shortest Path to Get All Keys (LC #864) | `shortest_path_get_all_keys.py` | BFS with state (keys collected) |
| Sliding Puzzle (LC #773) | `sliding_puzzle.py` | BFS in state space |

**🎓 Interview Tips:**
- BFS for shortest path (unweighted), DFS for exploration
- Use visited set to avoid cycles
- For state space search, use BFS to guarantee shortest path
- Track state as (position, additional_info) for complex problems


### Pattern 3: Shortest Path (Weighted Graphs)

**When to Use:**
- Weighted graphs
- Finding cheapest/fastest routes
- With constraints (like K stops)

**Dijkstra (Greedy - Non-negative weights):**
```python
import heapq
heap = [(0, start)]  # (cost, node)
dist = {start: 0}
while heap:
    cost, node = heapq.heappop(heap)
    if node == target: return cost
    for neighbor, weight in graph[node]:
        new_cost = cost + weight
        if new_cost < dist.get(neighbor, float('inf')):
            dist[neighbor] = new_cost
            heapq.heappush(heap, (new_cost, neighbor))
```

**Bellman-Ford (Handles negative weights, constraints):**
- For K stops constraint, use modified Bellman-Ford
- Relax edges K+1 times

**Problems:**
| Problem | File | Algorithm |
|---------|------|-----------|
| Cheapest Flights Within K Stops (LC #787) | `cheapest_flights_k_stops.py` | Modified Dijkstra or Bellman-Ford |

**🎓 Interview Tips:**
- Dijkstra for non-negative weights (O(E log V))
- Bellman-Ford for negative weights or constraints (O(VE))
- For K stops: track (node, stops_used) as state
- Use priority queue for Dijkstra, regular queue for Bellman-Ford

### Pattern 1: Sliding Window

**When to Use:**
- Contiguous substring/subarray problems
- Fixed or variable window size
- Longest/shortest substring with constraint

**Template:**
```python
left = 0
for right in range(len(s)):
    # Expand window
    # Update state
    while condition_not_met:
        # Shrink window from left
        left += 1
    # Update answer
```

# 4. Tree Problems ⭐⭐ **HIGH PRIORITY**

## 🎯 Why Airbnb Loves Trees
- Hierarchical data (categories, locations)
- File systems
- Decision trees
- LCA problems

## 💡 Tree Pattern Recognition

**Key Questions:**
1. Need to traverse? → DFS (pre/in/post) or BFS (level-order)
2. Need optimal solution? → Tree DP (bottom-up)
3. Need common ancestor? → LCA algorithm
4. Need to serialize? → Pre-order with null markers


### Pattern 1: Tree Traversal

**DFS Templates:**
```python
# Pre-order: root -> left -> right
def preorder(root):
    if not root: return
    process(root)
    preorder(root.left)
    preorder(root.right)

# In-order: left -> root -> right (BST gives sorted order)
# Post-order: left -> right -> root (useful for deletion)
```

**BFS Template (Level-order):**
```python
from collections import deque
queue = deque([root])
while queue:
    level_size = len(queue)
    for _ in range(level_size):
        node = queue.popleft()
        process(node)
        if node.left: queue.append(node.left)
        if node.right: queue.append(node.right)
```

**Problems:**
| Problem | File | Traversal Type |
|---------|------|----------------|
| Flatten Nested List Iterator (LC #341) | `nested_list_iterator.py` | DFS with stack |

**🎓 Interview Tips:**
- DFS: Recursive (cleaner) or Iterative (with stack)
- BFS: Always iterative with queue
- Know when to use each: DFS for paths, BFS for levels
- Time: O(n), Space: O(h) for DFS, O(w) for BFS (w=width)


### Pattern 2: Tree DP (Bottom-Up)

**When to Use:**
- Optimal solution depends on subtrees
- Need to return multiple values from subtree
- House Robber on tree

**Template:**
```python
def dfs(node):
    if not node: return base_case
    left_result = dfs(node.left)
    right_result = dfs(node.right)
    # Combine results from subtrees
    return combine(left_result, right_result, node)
```

**Problems:**
| Problem | File | Pattern |
|---------|------|---------|
| House Robber III (LC #337) | `house_robber_iii.py` | Return (rob, skip) for each node |
| Lowest Common Ancestor (LC #236) | `lowest_common_ancestor.py` | Return node if found, None otherwise |

**🎓 Interview Tips:**
- Think bottom-up: solve subtrees first
- Return multiple values if needed (tuple)
- Base case: handle None nodes
- Time: O(n), Space: O(h) for recursion

# 💡 Essential Interview Tips by Algorithm

## Dynamic Programming
1. **Always start with recursion** → then add memoization → then convert to bottom-up
2. **Identify the state** clearly: "What does dp[i] represent?"
3. **Find the recurrence**: How to build current from previous states?
4. **Base cases**: Handle smallest subproblems
5. **Space optimization**: Can you use O(1) or O(min(m,n)) instead of O(n) or O(m×n)?
6. **Mention time/space trade-offs** during interview

## Graph Algorithms
1. **Recognize patterns**: Dependencies → Topological Sort, Shortest path → BFS/Dijkstra
2. **Build graph correctly**: Adjacency list + in-degrees for topological sort
3. **Handle cycles**: Always check for cycles in topological sort
4. **State space search**: Use BFS to guarantee shortest path, track state as (node, info)
5. **Visited tracking**: Use set for O(1) lookup

## String Manipulation
1. **Sliding Window**: Expand right, shrink left when constraint violated
2. **Two Pointers**: Start from both ends for sorted arrays, handle different lengths
3. **Parsing**: Use state machine for complex parsing, stack for nested structures
4. **Trie**: Build in O(W×L), search in O(L), combine with DFS for grid search

## Tree Problems
1. **DFS vs BFS**: DFS for paths/exploration, BFS for levels/shortest path
2. **Tree DP**: Think bottom-up, return multiple values if needed
3. **LCA**: If both nodes in different subtrees → current is LCA
4. **Iterative vs Recursive**: Know both, mention trade-offs

## Design Problems
1. **Identify critical operations**: Optimize the most frequent ones
2. **Combine data structures**: HashMap + DLL, HashMap + Binary Search, etc.
3. **Mention scalability**: What if data grows? How to handle?
4. **Trade-offs**: Time vs space, read vs write performance
