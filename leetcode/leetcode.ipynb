```python
class Solution:
    def findSmallestRegion(self, regions, region1, region2):
        # Step 1: Build child -> parent map
        parent = {}
        for group in regions:
            p = group[0]
            for child in group[1:]:
                parent[child] = p

        # Step 2: Collect all ancestors of region1
        ancestors = set()
        cur = region1
        while cur:
            ancestors.add(cur)
            cur = parent.get(cur)


        # Step 3: Walk upward from region2 until match
        cur = region2
        while cur not in ancestors:
            cur = parent.get(cur)

        return cur

a=Solution()
regions = [
  ["Earth", "North America", "South America"],
  ["North America", "United States", "Canada"],
  ["United States", "New York", "Boston"]
]

region1 = "New York"
region2 = "Canada"

a.findSmallestRegion(regions, region1, region2)
```

```python
class FileSystem:

    def __init__(self):
        # store full paths
        self.paths = {}

    def createPath(self, path: str, value: int) -> bool:
        if path in self.paths:
            return False  # path already exists

        # find parent path
        parent = path[:path.rfind("/")]
        print(parent)
        if parent and parent not in self.paths:
            return False  # parent doesn't exist

        self.paths[path] = value
        return True

    def get(self, path: str) -> int:
        return self.paths.get(path, -1)
```
